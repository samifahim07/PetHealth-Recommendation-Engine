## Feature Enginnering

```python
# 1. Combine all symptoms into one text column
df["combined_symptoms"] = (
    df["symptom_1"].fillna("") + " " +
    df["symptom_2"].fillna("") + " " +
    df["symptom_3"].fillna("")
).str.strip()


# 2. Create age group
def age_group(age):
    if age < 1:
        return "Puppy/Kitten"
    elif age < 3:
        return "Young"
    elif age < 8:
        return "Adult"
    else:
        return "Senior"

df["age_group"] = df["age"].apply(age_group)


# 3. Severity score
severity_map = {
    "Mild": 1,
    "Moderate": 2,
    "Severe": 3
}

df["severity_score"] = df["severity"].map(severity_map)


# 4. Convert duration from hours to days
df["duration_days"] = df["duration"] / 24


# 5. Count how many symptoms are available
df["symptom_count"] = df[
    ["symptom_1", "symptom_2", "symptom_3"]
].notna().sum(axis=1)


# 6. Emergency symptom flag
emergency_keywords = [
    "difficulty breathing",
    "collapse",
    "seizure",
    "unable to urinate",
    "no urine",
    "toxin",
    "poison",
    "severe bleeding",
    "unresponsive"
]

df["emergency_flag"] = df["combined_symptoms"].str.lower().apply(
    lambda text: 1 if any(keyword in text for keyword in emergency_keywords) else 0
)


# 7. Severity + duration interaction feature
df["severity_duration_score"] = (
    df["severity_score"] * df["duration_days"]
)


# Check new features
df[
    [
        "pet_type",
        "age",
        "age_group",
        "combined_symptoms",
        "duration",
        "duration_days",
        "severity",
        "severity_score",
        "symptom_count",
        "emergency_flag",
        "severity_duration_score"
    ]
].head()
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Features
X = df[
    [
        "pet_type",
        "age",
        "age_group",
        "combined_symptoms",
        "duration",
        "duration_days",
        "severity",
        "severity_score",
        "symptom_count",
        "emergency_flag",
        "severity_duration_score"
    ]
]

# Targets
y_concern = df["possible_concern"]
y_urgency = df["urgency"]

# Group column
groups = df["scenario_group_id"]


# Leakage-safe split
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(X, y_concern, groups=groups)
)


# Train data
X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_concern_train = y_concern.iloc[train_idx]
y_concern_test = y_concern.iloc[test_idx]

y_urgency_train = y_urgency.iloc[train_idx]
y_urgency_test = y_urgency.iloc[test_idx]
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("Concern Train:", y_concern_train.shape)
print("Concern Test :", y_concern_test.shape)

print("Urgency Train:", y_urgency_train.shape)
print("Urgency Test :", y_urgency_test.shape)
```

```python
train_groups = set(groups.iloc[train_idx])
test_groups = set(groups.iloc[test_idx])

common_groups = train_groups.intersection(test_groups)

print("Common scenario groups:", common_groups)
print("Number of common groups:", len(common_groups))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer

# Text feature
text_feature = "combined_symptoms"

# Categorical features
categorical_features = [
    "pet_type",
    "age_group",
    "severity"
]

# Numerical features
numerical_features = [
    "age",
    "duration",
    "duration_days",
    "severity_score",
    "symptom_count",
    "severity_duration_score"
]

# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        
        # Symptoms -> TF-IDF
        (
            "symptoms",
            TfidfVectorizer(
                lowercase=True,
                ngram_range=(1, 2)
            ),
            text_feature
        ),

        # Categorical -> One Hot Encoding
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),

        # Numerical -> Scaling
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
X_train_transformed = preprocessor.fit_transform(X_train)

print("Original shape:", X_train.shape)
print("Transformed shape:", X_train_transformed.shape)
```

```python
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import ExtraTreesClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import pandas as pd


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Linear SVM": LinearSVC(
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "LightGBM": LGBMClassifier(
        random_state=42,
        verbose=-1
    )
}


results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    # Train
    pipeline.fit(X_train, y_concern_train)

    # Predict
    y_pred = pipeline.predict(X_test)

    # Evaluation
    accuracy = accuracy_score(y_concern_test, y_pred)

    precision = precision_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "Macro F1": f1
    })


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Macro F1",
    ascending=False
)

results_df
```

```python
def map_concern(concern):
    concern = concern.lower()

    if any(word in concern for word in ["digestive", "gastro", "abdominal"]):
        return "Digestive"

    elif "respiratory" in concern:
        return "Respiratory"

    elif "urinary" in concern:
        return "Urinary"

    elif any(word in concern for word in ["skin", "allergy", "allergic"]):
        return "Skin/Allergy"

    elif "ear" in concern:
        return "Ear"

    elif "eye" in concern:
        return "Eye"

    elif any(word in concern for word in ["mobility", "injury", "trauma", "bleeding"]):
        return "Mobility/Injury"

    elif "neurolog" in concern:
        return "Neurological"

    elif any(word in concern for word in ["toxic", "poison"]):
        return "Toxic Exposure"

    elif any(word in concern for word in ["dental", "oral"]):
        return "Dental/Oral"

    elif "parasite" in concern:
        return "Parasite"

    else:
        return "General/Emergency"


df["concern_category"] = df["possible_concern"].apply(map_concern)
```

```python
print(df["concern_category"].value_counts())

print("\nTotal Categories:")
print(df["concern_category"].nunique())
```

```python
from sklearn.model_selection import StratifiedGroupKFold

X = df[
    [
        "pet_type",
        "age",
        "age_group",
        "combined_symptoms",
        "duration",
        "duration_days",
        "severity",
        "severity_score",
        "symptom_count",
        "severity_duration_score"
    ]
]

y_concern = df["concern_category"]
groups = df["scenario_group_id"]


sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_idx, test_idx = next(
    sgkf.split(X, y_concern, groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_concern_train = y_concern.iloc[train_idx]
y_concern_test = y_concern.iloc[test_idx]
```

```python
train_groups = set(groups.iloc[train_idx])
test_groups = set(groups.iloc[test_idx])

common_groups = train_groups.intersection(test_groups)

print("Common groups:", common_groups)
print("Number of common groups:", len(common_groups))
```

```python
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import ExtraTreesClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import pandas as pd


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Linear SVM": LinearSVC(
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "LightGBM": LGBMClassifier(
        random_state=42,
        verbose=-1
    )
}


results = []
trained_models = {}


for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    # Train
    pipeline.fit(X_train, y_concern_train)

    # Predict
    y_pred = pipeline.predict(X_test)

    # Save trained model
    trained_models[name] = pipeline

    # Evaluation
    accuracy = accuracy_score(y_concern_test, y_pred)

    precision = precision_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "Macro F1": f1
    })


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Macro F1",
    ascending=False
)

results_df
```

```python
from sklearn.metrics import classification_report

best_concern_model = trained_models["Linear SVM"]

y_pred_concern = best_concern_model.predict(X_test)

print(
    classification_report(
        y_concern_test,
        y_pred_concern,
        zero_division=0
    )
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

fig, ax = plt.subplots(figsize=(12, 10))

ConfusionMatrixDisplay.from_predictions(
    y_concern_test,
    y_pred_concern,
    xticks_rotation=90,
    cmap="Blues",
    ax=ax
)

plt.title("Linear SVM - Concern Prediction")
plt.tight_layout()
plt.show()
```
