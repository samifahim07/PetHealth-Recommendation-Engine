## Feature Enginnering

```python
# 1. Combine all symptoms into one text column
df["combined_symptoms"] = (
    df["symptom_1"].fillna("") + " " +
    df["symptom_2"].fillna("") + " " +
    df["symptom_3"].fillna("")
).str.strip()


# 2. Create age group
def age_group(age):
    if age < 1:
        return "Puppy/Kitten"
    elif age < 3:
        return "Young"
    elif age < 8:
        return "Adult"
    else:
        return "Senior"

df["age_group"] = df["age"].apply(age_group)


# 3. Severity score
severity_map = {
    "Mild": 1,
    "Moderate": 2,
    "Severe": 3
}

df["severity_score"] = df["severity"].map(severity_map)


# 4. Convert duration from hours to days
df["duration_days"] = df["duration"] / 24


# 5. Count how many symptoms are available
df["symptom_count"] = df[
    ["symptom_1", "symptom_2", "symptom_3"]
].notna().sum(axis=1)


# 6. Emergency symptom flag
emergency_keywords = [
    "difficulty breathing",
    "collapse",
    "seizure",
    "unable to urinate",
    "no urine",
    "toxin",
    "poison",
    "severe bleeding",
    "unresponsive"
]

df["emergency_flag"] = df["combined_symptoms"].str.lower().apply(
    lambda text: 1 if any(keyword in text for keyword in emergency_keywords) else 0
)


# 7. Severity + duration interaction feature
df["severity_duration_score"] = (
    df["severity_score"] * df["duration_days"]
)


# Check new features
df[
    [
        "pet_type",
        "age",
        "age_group",
        "combined_symptoms",
        "duration",
        "duration_days",
        "severity",
        "severity_score",
        "symptom_count",
        "emergency_flag",
        "severity_duration_score"
    ]
].head()
```

## Neurological and Toxic Exposure Feature Flags


```python
neurological_keywords = [
    "seizure",
    "paralysis",
    "confusion",
    "loss of balance",
    "unresponsive",
    "inability to walk"
]

df["neurological_flag"] = df["combined_symptoms"].str.lower().apply(
    lambda text: 1 if any(keyword in text for keyword in neurological_keywords) else 0
)

toxic_keywords = [
    "toxin",
    "poison",
    "toxic ingestion",
    "chemical",
    "suspected toxin ingestion"
]

df["toxic_flag"] = df["combined_symptoms"].str.lower().apply(
    lambda text: 1 if any(keyword in text for keyword in toxic_keywords) else 0
)

df[["combined_symptoms", "neurological_flag", "toxic_flag"]].head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer

text_feature = "combined_symptoms"

categorical_features = [
    "pet_type",
    "age_group",
    "severity"
]

numerical_features = [
    "age",
    "duration",
    "duration_days",
    "severity_score",
    "symptom_count",
    "severity_duration_score",
    "neurological_flag",
    "toxic_flag"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "symptoms",
            TfidfVectorizer(
                lowercase=True,
                ngram_range=(1, 2)
            ),
            text_feature
        ),
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)

print("Updated preprocessing pipeline created successfully.")
```

```python
def map_concern(concern):
    concern = concern.lower()

    if any(word in concern for word in ["digestive", "gastro", "abdominal"]):
        return "Digestive"

    elif "respiratory" in concern:
        return "Respiratory"

    elif "urinary" in concern:
        return "Urinary"

    elif any(word in concern for word in ["skin", "allergy", "allergic"]):
        return "Skin/Allergy"

    elif "ear" in concern:
        return "Ear"

    elif "eye" in concern:
        return "Eye"

    elif any(word in concern for word in ["mobility", "injury", "trauma", "bleeding"]):
        return "Mobility/Injury"

    elif "neurolog" in concern:
        return "Neurological"

    elif any(word in concern for word in ["toxic", "poison"]):
        return "Toxic Exposure"

    elif any(word in concern for word in ["dental", "oral"]):
        return "Dental/Oral"

    elif "parasite" in concern:
        return "Parasite"

    else:
        return "General/Emergency"


df["concern_category"] = df["possible_concern"].apply(map_concern)
```

```python
from sklearn.model_selection import StratifiedGroupKFold

X = df[
    [
        "pet_type",
        "age",
        "age_group",
        "combined_symptoms",
        "duration",
        "duration_days",
        "severity",
        "severity_score",
        "symptom_count",
        "severity_duration_score",
        "neurological_flag",
        "toxic_flag"
    ]
]

y_concern = df["concern_category"]
groups = df["scenario_group_id"]

sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_idx, test_idx = next(
    sgkf.split(X, y_concern, groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_concern_train = y_concern.iloc[train_idx]
y_concern_test = y_concern.iloc[test_idx]
```

```python
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import ExtraTreesClassifier
from lightgbm import LGBMClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import pandas as pd


models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Linear SVM": LinearSVC(
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "LightGBM": LGBMClassifier(
        random_state=42,
        verbose=-1
    )
}


results = []
trained_models = {}


for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    # Train
    pipeline.fit(X_train, y_concern_train)

    # Predict
    y_pred = pipeline.predict(X_test)

    # Save trained model
    trained_models[name] = pipeline

    # Evaluation
    accuracy = accuracy_score(y_concern_test, y_pred)

    precision = precision_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_concern_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "Macro F1": f1
    })


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Macro F1",
    ascending=False
)

results_df
```

```python
def get_recommendation(
    predicted_concern,
    predicted_urgency,
    symptoms
):

    candidates = df[
        (df["concern_category"] == predicted_concern) &
        (df["urgency"] == predicted_urgency)
    ].copy()

    # If exact urgency + concern combination not found
    if len(candidates) == 0:
        candidates = df[
            df["concern_category"] == predicted_concern
        ].copy()

    if len(candidates) == 0:
        return {
            "possible_concern": predicted_concern,
            "basic_action": "Veterinary assessment recommended.",
            "red_flags": "Monitor for worsening symptoms."
        }

    input_words = set(symptoms.lower().split())

    def symptom_similarity(row):

        row_text = (
            str(row["symptom_1"]) + " " +
            str(row["symptom_2"]) + " " +
            str(row["symptom_3"])
        ).lower()

        row_words = set(row_text.split())

        return len(
            input_words.intersection(row_words)
        )

    candidates["similarity"] = candidates.apply(
        symptom_similarity,
        axis=1
    )

    best_match = candidates.sort_values(
        by="similarity",
        ascending=False
    ).iloc[0]

    return {
        "possible_concern": best_match["possible_concern"],
        "basic_action": best_match["basic_action"],
        "red_flags": best_match["red_flags"]
    }
```

```python
symptom_columns = [
    "symptom_1",
    "symptom_2",
    "symptom_3"
]

symptom_vocabulary = set()

for column in symptom_columns:

    values = (
        df[column]
        .dropna()
        .astype(str)
        .str.lower()
        .unique()
    )

    symptom_vocabulary.update(values)

symptom_vocabulary = sorted(
    symptom_vocabulary,
    key=len,
    reverse=True
)

print("Total Known Symptoms:", len(symptom_vocabulary))
```

```python
import re

def extract_pet_information(text):

    text_lower = text.lower()

    # Pet type
    if "cat" in text_lower:
        pet_type = "Cat"

    elif "dog" in text_lower:
        pet_type = "Dog"

    else:
        pet_type = None


    # Symptoms
    symptoms = []

    for symptom in symptom_vocabulary:

        if symptom in text_lower:
            symptoms.append(symptom)


    symptoms = list(dict.fromkeys(symptoms))


    # Duration
    duration_hours = None

    hour_match = re.search(
        r"(\d+)\s*(hour|hours|hr|hrs)",
        text_lower
    )

    day_match = re.search(
        r"(\d+)\s*(day|days)",
        text_lower
    )

    if hour_match:

        duration_hours = int(
            hour_match.group(1)
        )

    elif day_match:

        duration_hours = (
            int(day_match.group(1)) * 24
        )


    return {
        "pet_type": pet_type,
        "symptoms": symptoms,
        "duration_hours": duration_hours
    }
```

```python
severity_map = {
    "Mild": 1,
    "Moderate": 2,
    "Severe": 3
}
```

```python
def create_special_flags(symptom_text):

    text = symptom_text.lower()

    neurological_flag = int(
        any(
            keyword in text
            for keyword in neurological_keywords
        )
    )

    toxic_flag = int(
        any(
            keyword in text
            for keyword in toxic_keywords
        )
    )

    return neurological_flag, toxic_flag
```

```python
def predict_pet_health(
    pet_type,
    age,
    symptom_1,
    symptom_2,
    symptom_3,
    duration,
    severity
):

    combined_symptoms = " ".join([
        symptom_1,
        symptom_2,
        symptom_3
    ]).strip()

    age_group = get_age_group(age)

    severity_score = severity_map[severity]

    duration_days = duration / 24

    symptom_count = sum([
        bool(symptom_1),
        bool(symptom_2),
        bool(symptom_3)
    ])

    severity_duration_score = (
        severity_score * duration_days
    )

    neurological_flag, toxic_flag = (
        create_special_flags(
            combined_symptoms
        )
    )


    user_data = pd.DataFrame([{
        "pet_type": pet_type,
        "age": age,
        "age_group": age_group,
        "combined_symptoms": combined_symptoms,
        "duration": duration,
        "duration_days": duration_days,
        "severity": severity,
        "severity_score": severity_score,
        "symptom_count": symptom_count,
        "severity_duration_score":
            severity_duration_score,
        "neurological_flag":
            neurological_flag,
        "toxic_flag":
            toxic_flag
    }])


    # Concern prediction
    predicted_concern = (
        final_concern_model.predict(
            user_data
        )[0]
    )


    # Urgency prediction
    predicted_urgency = (
        final_urgency_model.predict(
            user_data
        )[0]
    )


    # Emergency safety layer
    final_urgency = emergency_override(
        combined_symptoms,
        predicted_urgency
    )


    # Recommendation
    recommendation = get_recommendation(
        predicted_concern,
        final_urgency,
        combined_symptoms
    )


    return {
        "Concern Category":
            predicted_concern,

        "Possible Concern":
            recommendation[
                "possible_concern"
            ],

        "ML Urgency":
            predicted_urgency,

        "Final Urgency":
            final_urgency,

        "Basic Action":
            recommendation[
                "basic_action"
            ],

        "Red Flags":
            recommendation[
                "red_flags"
            ]
    }
```

```python
result = predict_pet_health(
    pet_type="Cat",
    age=3,
    symptom_1="vomiting",
    symptom_2="lethargy",
    symptom_3="reduced appetite",
    duration=24,
    severity="Moderate"
)

for key, value in result.items():
    print(f"{key}: {value}")
```
